```python
# 1. Check whether input number is even or odd. Handle invalid input using exception handling.
try:
    num = int(input("Enter a number"))
    if num%2 ==0:
        print("Even Number:-",num)
    else:
        print("Odd Number:-",num) 
except Exception as e:
    print("Exception is:-",e)           
```

```python
# 2. Take a number from user. If input is valid: (using math build-in function)
# Print ceil if number > 0
# Print floor if number < 0
# Handle invalid input

import math

try:
    num = float(input("Enter a number: "))
    
    if num > 0:
        print("Ceil:", math.ceil(num))
    elif num < 0:
        print("Floor:", math.floor(num))
    else:
        print("Number is zero")

except Exception as e:
    print("Exception is:", e)

print("ALL OKK")       
```

```python
# 3. Write a program to divide two numbers entered by user. Handle the case where the denominator is zero. Print the result if no error is found.
#  using try-except-else-finally

try:
    a = int(input("Enter a number"))
    b = int(input("Enter a number"))
    result = a/b
except ZeroDivisionError:
    print("Exception is:- denominator is zero")
except ValueError:
    print("Invalid input")
else:
    print("Result is:-",result) 
finally:
    print("End")       
```

```python
#4. Write a program to convert a string input into an integer. Handle invalid input.
try:
    s = input("Enter a string") 
    a = int(s)
    print("Converted integer:-",a)
except ValueError:
    print("Invalid input! Cannot convert to integer.")  
  
 
```
