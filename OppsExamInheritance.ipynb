```python
#1. Create a Vehicle class with attributes brand and speed. Create a child class Car that inherits from Vehicle and adds fuel_type.

class Vehicle():
    def __init__(self,brand,speed):
        self.brand=brand
        self.speed=speed


class Car(Vehicle):
    def __init__(self,brand,speed,fuel):
        super().__init__(brand,speed)

        self.fuel=fuel

c1 = Car("Toyota",180,"Petrol")

print("Barnd:-",c1.brand)
print("Speed:-",c1.speed)
print("Fuel type:-",c1.fuel)
```

```python
# 2. Create a Person class with attributes name and age. Create a subclass Student with an additional attribute grade.

class Person():
    def __init__(self,name,age):
        self.name=name
        self.age=age

class Student(Person):
    def __init__(self,name,age,location,grade):
        super().__init__(name,age)
        self.location = location
        self.hobby = grade

s1 = Student("Virat",30,"London","A") 

print("Name:-",s1.name)
print("Age:-",s1.age)
print("Location:-",s1.location)
print("Grade:-",s1.hobby)
```

```python
# 3. Create a Shape class with a method area(). Create subclasses Rectangle and Circle that calculate their respective areas.

class Shape():
    def area(self):
        pass
        

class Rectangle(Shape):
    def __init__(self,length,width):
        self.length=length
        self.width=width

    def area(self):
        return self.length * self.width   
        

class Circle(Shape):
    def __init__(self,radius,pi=3.14):
        self.radius = radius
        self.pi=pi

    def area(self):
        return self.pi * (self.radius**2)
        
    
a1 = Rectangle(10,8)

print("Area of Rectangle is :-",a1.area()) 

a2 = Circle(10)
print("Area of Circle is:-",a2.area())
```

```python
# 5. Create a BankAccount class with attributes account_number and balance.
# Create a subclass SavingsAccount with an additional interest rate.

class BankAccount():
    def __init__(self,account_number,balance):
        self.account_number=account_number
        self.balance = balance

class SavingAccount(BankAccount):
    def __init__(self,account_number,balance,interest_rate):
        super().__init__(account_number,balance) 
        self.interest_rate = interest_rate

s1 = SavingAccount(53447811001,700000,4.2)
print("Account Number:-",s1.account_number) 
print("Balance:-",s1.balance)
print("Interest Rate:-",s1.interest_rate)              
```

```python
# 6. Create a class Animal, then Mammal inheriting from Animal, and Dog inheriting from Mammal. Add methods in each class

class Animal():
    def __init__(self,name):
        self.name = name

    def show(self):
        print("Animal Name:-",self.name)    

class Mammal(Animal):
    def mammal_type(self):
        print("This is Mammal.")
    

class Dog(Mammal):
    def sound(self):
        print("Dog barks")

d1 = Dog("Tommy")

d1.show()
d1.mammal_type()
d1.sound()        
```

```python
# 7. Create a class Employee, then Manager inheriting from Employee, and SeniorManager inheriting from Manager.

class Employee():
    def __init__(self,name,department):
        self.name=name
        self.department=department 

class Manager(Employee):
    def __init__(self,name,department,salary):
        super().__init__(name,department)
        self.salary = salary
        

class SeniorManager(Manager):
    def __init__(self,name,department,salary,bonus):
        super().__init__(name,department,salary)
        self.bonus = bonus

e1 = SeniorManager("Skashi","HR",60000,25000) 
print(f"Name:{e1.name},Department:{e1.department},Salary:{e1.salary},Bonus:{e1.bonus}")
```

```python
# 8. Create a Vehicle class → Car class → ElectricCar class with additional battery capacity.

class Vehicle():
    def __init__(self,name):
        self.name = name

class Car(Vehicle):
    def __init__(self,name,brand,model):
        super().__init__(name)
        self.brand=brand
        self.model=model

class ElectricCar(Car):
    def __init__(self,name,brand,model,price,battery):
        super().__init__(name,brand,model)
        self.price=price
        self.battery=battery

e1 = ElectricCar("Kia Syros","KIA","EV5",2000000,"42 KWH") 
print(f"Brand:{e1.brand},Name:{e1.name},Model:{e1.model},Price:{e1.price},BatteryBackup:{e1.battery}")       
```

```python
# 9. Create a Person class → Employee class → Developer class with programming 

class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name, city, joining_date):
        super().__init__(name)
        self.city = city
        self.joining_date = joining_date

class Developer(Employee):
    def __init__(self, name, city, joining_date, language):
        super().__init__(name, city, joining_date)
        self.language = language

# Object
d1 = Developer("Virat", "Pune", "1 April 2026", "Python and Java")

print("Name:-", d1.name)
print("City:-", d1.city)
print("Joining Date:-", d1.joining_date)
print("Language:-", d1.language)
```

```python
# 10. Create a class hierarchy for University → Department → Course.

class University:
    def __init__(self,name):
        self.name=name

class Department(University):
    def __init__(self,name,dep_name):
        super().__init__(name)
        self.dep_name=dep_name

class Course (Department):
    def __init__(self,name,dep_name,cour_name):
        super().__init__(name,dep_name)
        self.cour_name=cour_name

c1 = Course("Pune University","Deparment of Management Science","MBA")    

print("University:-",c1.name)
print("Department:-",c1.dep_name)
print("Course:-",c1.cour_name)
```
