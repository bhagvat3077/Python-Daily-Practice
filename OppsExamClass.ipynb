```python
# CLASS Compnay,attributes=name,sal,dept.....
# sal() print salary
# fun() named = bonus=salary 5% add.
class Company:
    def __init__(self,name,sal,dept):
        self.name=name
        self.sal=sal
        self.dept=dept

    def bonus(self):
        t=0.05  # extra bonous
        total = self.sal*t
        final_sal = self.sal+total
        print("Yours salary is:-",self.sal,"\n","your Bonus:-",total,"\n","Total Salary is:-",final_sal)
        

emp1 = Company("Akash",100000,"HR") 
print(emp1.name)
print(emp1.sal)
print(emp1.dept)
emp1.bonus()       

        
```

```python
# 3. Write a class Car that takes brand and model as parameters using the init method and prints them using another method display().

class Car():

    def __init__(self,brand,model):
        self.brand=brand
        self.model=model

    def display(self):
        print("Brand is:-",self.brand) 
        print("Model is:-",self.model)


c1 = Car("TATA","Safari")
c1.display()

c2 = Car("KIA","Seltos")
c2.display()

```

```python
# 5. Write a class BankAccount with attributes account_holder and balance. 
# Add a method deposit(amount) that increases the balance.

class BankAccount():
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.balance=balance

    def deposit(self,amount):
        self.balance = self.balance+amount
        print("After Deposit, Your Net Balance is :-",self.balance)

b1 = BankAccount("Virat Sharma",100000)
print("Your Name:-",b1.account_holder)
print("Your Balance:-",b1.balance)
b1.deposit(5000)
```

```python
# 7. Write a class Rectangle that takes length and width in init and has a method area() that returns the area.

class Rectangle():
    def __init__(self,length,width):
        self.length=length
        self.width=width

    def area(self):
        a=self.length*self.width
        print("Area :-",a)

r1 = Rectangle(7,10)
r1.area()
```

```python
# 10. Write a class Counter that counts how many objects have been created using a class variable.

class Counter ():
    count = 0

    def __init__(self):
        Counter.count = Counter.count+1

c1 = Counter()
c2 = Counter()
c3 = Counter()

print("Total Object Created:-",Counter.count)
```

```python
# 5. Create a static method to check if a number is even.
class Even():

    @staticmethod
    def check_no(num):
        if num%2 == 0:
            print("Even Number:-",num)
        else:
            print("Odd Number:-",num) 
Even.check_no(7)  
Even.check_no(20)             
```

```python
# 6. Override __str__() to print object details.

class Student():
    def __init__(self,name,age):
        self.name=name
        self.age=age
        
    def __str__(self):
        return f"{self.name} {self.age}"    

p1 = Student("Arjun",30)
print(p1)        
```
