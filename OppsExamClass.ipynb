```python
# CLASS Compnay,attributes=name,sal,dept.....
# sal() print salary
# fun() named = bonus=salary 5% add.
class Company:
    def __init__(self,name,sal,dept):
        self.name=name
        self.sal=sal
        self.dept=dept

    def bonus(self):
        t=0.05  # extra bonous
        total = self.sal*t
        final_sal = self.sal+total
        print("Yours salary is:-",self.sal,"\n","your Bonus:-",total,"\n","Total Salary is:-",final_sal)
        

emp1 = Company("Akash",100000,"HR") 
print(emp1.name)
print(emp1.sal)
print(emp1.dept)
emp1.bonus()       

        
```

```python
# 3. Write a class Car that takes brand and model as parameters using the init method and prints them using another method display().

class Car():

    def __init__(self,brand,model):
        self.brand=brand
        self.model=model

    def display(self):
        print("Brand is:-",self.brand) 
        print("Model is:-",self.model)


c1 = Car("TATA","Safari")
c1.display()

c2 = Car("KIA","Seltos")
c2.display()

```

```python
#4. Write a class Employee that stores employee name and salary using init. Create two objects and display their details.

class Employee():
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary

e1 = Employee("Virat",50000) 
print("Name:-",e1.name) 
print("Salary:-",e1.salary) 

e2 = Employee("Akash",70000)
print("Name:-",e2.name) 
print("Salary:-",e2.salary) 
```

```python
# 5. Write a class BankAccount with attributes account_holder and balance. 
# Add a method deposit(amount) that increases the balance.

class BankAccount():
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.balance=balance

    def deposit(self,amount):
        self.balance = self.balance+amount
        print("After Deposit, Your Net Balance is :-",self.balance)

b1 = BankAccount("Virat Sharma",100000)
print("Your Name:-",b1.account_holder)
print("Your Balance:-",b1.balance)
b1.deposit(5000)
```
